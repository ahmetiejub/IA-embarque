```python
import tensorflow as tf, subprocess, sys
m = tf.keras.Sequential([tf.keras.Input((784,)), tf.keras.layers.Dense(10, activation="softmax")])
m.export("test_sm")
r = subprocess.run([sys.executable, "-m", "tf2onnx.convert", "--saved-model", "test_sm",
                    "--output", "test.onnx", "--opset", "13"], capture_output=True, text=True)
print("OK export ONNX" if r.returncode == 0 else r.stderr[-1500:])
```
